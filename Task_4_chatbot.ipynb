```python

import re
import json

# Sample FAQ Dataset
faq_data = """
{
    "business_hours": "Monday to Saturday, 9 AM to 6 PM",
    "return_policy": "Products can be returned within 7 days.",
    "products": {
        "mouse": "Wireless Mouse - Price: Rs. 500",
        "keyboard": "Keyboard - Price: Rs. 800",
        "headphones": "Headphones - Price: Rs. 1000"
    }
}
"""

faq = json.loads(faq_data)

print("Welcome to Python Customer Support Chatbot!")
print("You can ask about business hours, products and returns.")
print("Type 'bye', 'exit' or 'quit' to stop.")

while True:

    user = input("\nYou: ").lower().strip()

    # Exit command
    if re.fullmatch(r"(bye|exit|quit)", user):
        print("Bot: Thank you! Have a nice day!")
        break

    # Greeting
    elif re.search(r"\b(hi|hello|hey)\b", user):
        print("Bot: Hello! How can I help you?")

    else:
        found = False

        # Business hours
        if re.search(r"\b(hours|timing|open|closing)\b", user):
            print("Bot: Our business hours are:",
                  faq["business_hours"])
            found = True

        # Return policy
        if re.search(r"\b(return|refund|exchange)\b", user):
            print("Bot: Return Policy:",
                  faq["return_policy"])
            found = True

        # Product details
        if re.search(r"\b(product|price|cost|mouse|keyboard|headphones)\b", user):

            for product, details in faq["products"].items():

                if re.search(r"\b" + product + r"\b", user):
                    print("Bot:", details)
                    found = True

        # Fallback response
        if not found:
            print("Bot: Sorry, I didn't understand.")
            print("Please ask about products, timings or returns.")
```
