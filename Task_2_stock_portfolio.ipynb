```python

import csv

# Sample stock prices (demo values only)
sample_prices = {
    "AAPL": 190,
    "MSFT": 420,
    "GOOGL": 175,
    "AMZN": 185
}

# Create sample stock price CSV file
with open("stock_prices.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Ticker", "Current_Price"])

    for ticker, price in sample_prices.items():
        writer.writerow([ticker, price])

# Read stock prices from CSV
current_prices = {}

with open("stock_prices.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        current_prices[row["Ticker"]] = float(row["Current_Price"])

portfolio = []

count = int(input("How many stocks do you want to enter? "))

for i in range(count):

    print("\nEnter Stock Details")

    ticker = input("Enter stock symbol: ").upper()

    buy_price = float(input("Enter buying price: "))

    shares = int(input("Enter number of shares: "))

    if ticker in current_prices:
        market_price = current_prices[ticker]
    else:
        market_price = float(input("Enter current market price: "))

    invested = buy_price * shares

    current_value = market_price * shares

    profit_loss = current_value - invested

    percentage = (profit_loss / invested) * 100 if invested else 0

    portfolio.append({
        "Ticker": ticker,
        "Buy_Price": buy_price,
        "Current_Price": market_price,
        "Shares": shares,
        "Invested_Amount": invested,
        "Current_Value": current_value,
        "Profit_Loss": profit_loss,
        "Percentage": percentage
    })

# Display portfolio
print("\n----- STOCK PORTFOLIO SUMMARY -----")

for stock in portfolio:
    print("\nStock:", stock["Ticker"])
    print("Invested:", stock["Invested_Amount"])
    print("Current Value:", stock["Current_Value"])
    print("Profit/Loss:", stock["Profit_Loss"])
    print("Gain/Loss %:", round(stock["Percentage"], 2))

# Calculate total
total_invested = sum(s["Invested_Amount"] for s in portfolio)
total_value = sum(s["Current_Value"] for s in portfolio)
total_profit = total_value - total_invested

print("\nTotal Investment:", total_invested)
print("Total Current Value:", total_value)
print("Total Profit/Loss:", total_profit)

# Save results into CSV
with open("portfolio_summary.csv", "w", newline="") as file:

    fieldnames = [
        "Ticker", "Buy_Price", "Current_Price",
        "Shares", "Invested_Amount", "Current_Value",
        "Profit_Loss", "Percentage"
    ]

    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(portfolio)

print("\nPortfolio saved successfully!")
print("File: portfolio_summary.csv")
```
